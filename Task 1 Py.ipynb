# Descriptive Statstics :

```python
# Performing the operation on the columns " AGE "

import pandas as pd
import statistics as st
df = pd.read_csv("retail_sales_dataset.csv")
mean_age = df["Age"].mean()
print("Mean of Age :",mean_age)

median_age = (df['Age'].median())
print("Median of Age :",median_age)

mode_age = (df['Age'].mode())
print("Mode of Age :",mode_age)

std_age = (df['Age'].std())
print("Standard Deviation of Age :",std_age)
```

```python
# Column - "Transaction ID"

mean_id = df['Transaction ID'].mean()
print("Mean of Transaction_Id :",mean_id)

median_id = (df['Transaction ID'].median())
print("Median of Transaction_Id :",median_id)

mode_id = (df['Transaction ID'].mode())
print("Mode of Transaction_Id :",mode_id)

std_id = (df['Transaction ID'].std())
print("Standard Deviaton of Transaction_Id :",std_id)
```

```python
# Column - "Quantity":

mean_qty = df['Quantity'].mean()
print("Mean of Quantity :",mean_qty)

median_qty = (df['Quantity'].median())
print("Median of Quantity :",median_qty)

mode_qty = (df['Quantity'].mode())
print("Mode of Quantity :",mode_qty)

std_qty = (df['Quantity'].std())
print("Standard Deviaton of Quantity :",std_qty)
```

```python
# Column - "Price Per Unit" :

mean_price_unit = df['Price per Unit'].mean()
print("Mean of Price per Unit :",mean_price_unit)

median_price_unit = (df['Price per Unit'].median())
print("Median of Price per Unit :",median_price_unit)

mode_price_unit = (df['Price per Unit'].mode())
print("Mode of Price per Unit:",mode_price_unit)

std_price_unit = (df['Price per Unit'].std())
print("Standard Deviaton of Price per Unit :",std_price_unit)
```

```python
# Column - " Total Amount ":

mean_total_amt = df['Total Amount'].mean()
print("Mean of Total Amount :",mean_total_amt)

median_total_amt = (df['Total Amount'].median())
print("Median of Total Amount :",median_total_amt)

mode_total_amt = (df['Total Amount'].mode())
print("Mode of Total Amount:",mode_total_amt)

std_total_amt = (df['Total Amount'].std())
print("Standard Deviaton of Total Amount :",std_total_amt)
```

# Time Series Analysis :

```python
# Convert text date column into datetime format

df["Date"] = pd.to_datetime(df["Date"])
print(df["Date"].dtype)
```

```python
#Finding the Monthly Sales using line charts

monthly_sales = df.resample("ME",on="Date")["Total Amount"].sum()
print(monthly_sales)
```

```python
# Monthly Sales Trend Using Plot

import pandas as pd
import matplotlib.pyplot as plt

df["Date"]=pd.to_datetime(df["Date"])
df=df.dropna(subset=["Date","Total Amount"])

monthly_sales = (
    df.resample("ME", on="Date")["Total Amount"]
      .sum()
      .reset_index()
)

plt.figure(figsize=(10,5))

plt.plot(
    monthly_sales["Date"],
    monthly_sales["Total Amount"],
    marker="o",
    color="blue",
    linewidth=0.5
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total_Sales_Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Finding the Quarterly Sales using line charts

Quarterly_Sales_Amount = df.resample("QE",on="Date")["Total Amount"].sum()
print(Quarterly_Sales_Amount)
```

```python
# Quaterly Sales Trend Plot

df.dropna(subset=["Date","Total Amount"])

Quarterly_Sales_Amount = ( 
                df.resample("QE",on="Date")["Total Amount"]
            .sum()
            .reset_index()
)

plt.figure(figsize=(8,5))

plt.plot(
     Quarterly_Sales_Amount["Date"],
    Quarterly_Sales_Amount["Total Amount"],
    marker="*",
    color="green",
    linewidth=1
)
plt.title("Quarterly Sales Trend")
plt.xlabel("Quarters")
plt.ylabel("Quarterly_Sales_Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# age-group distribution summary

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert required columns to numeric
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Total Amount"] = pd.to_numeric(df["Total Amount"], errors="coerce")

# Remove rows having missing values in required columns
df = df.dropna(subset=["Age", "Quantity", "Total Amount"])

# Create age groups
bins = [0, 18, 25, 35, 45, 55, 65, 100]

labels = [
    "Below 18",
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-65",
    "66+"
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Age-group summary
age_group_summary = (
    df.groupby("Age Group", observed=False)
      .agg(
          Total_Sales=("Total Amount", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Average_Age=("Age", "mean"),
          Customer_Count=("Customer ID", "nunique"),
          Transaction_Count=("Transaction ID", "nunique")
      )
      .reset_index()
)

print(age_group_summary)
```

# Product Analysis :

```python
import matplotlib.pyplot as plt

# Calculate values
revenue_by_category = (
    df.groupby("Product Category")["Total Amount"]
      .sum()
      .sort_values(ascending=False)
)

quantity_by_category = (
    df.groupby("Product Category")["Quantity"]
      .sum()
      .sort_values(ascending=False)
)

# Create one output with two charts
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Chart 1: Revenue by category
ax1 = revenue_by_category.plot(
    kind="bar",
    color="orange",
    ax=axes[0]
)

ax1.bar_label(
    ax1.containers[0],
    labels=[f"₹{value:,.0f}" for value in revenue_by_category.values],
    padding=3,
    fontsize=9
)

ax1.set_title("Revenue by Product Category")
ax1.set_xlabel("Product Category")
ax1.set_ylabel("Total Revenue")
ax1.tick_params(axis="x", rotation=0)
ax1.grid(axis="y", linestyle="--", alpha=0.5)


# Chart 2: Quantity by category
ax2 = quantity_by_category.plot(
    kind="bar",
    color="mediumseagreen",
    ax=axes[1]
)


ax2.bar_label(
    ax2.containers[0],
    labels=[f"{value:,.0f}" for value in quantity_by_category.values],
    padding=3,
    fontsize=9
)

ax2.set_title("Best-Selling Category by Quantity")
ax2.set_xlabel("Product Category")
ax2.set_ylabel("Total Quantity Sold")
ax2.tick_params(axis="x", rotation=0)
ax2.grid(axis="y", linestyle="--", alpha=0.5)

fig.suptitle("Product Category Analysis", fontsize=16, fontweight="bold")
plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.show()  
```

```python
revenue_by_category = (
    df.groupby("Product Category")["Total Amount"]
    .sum()
    .sort_values(ascending = False)
)

plt.figure(figsize=(7,5))

plt.pie(
    revenue_by_category.values,
    labels=revenue_by_category.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["pink","green","yellow"],
    wedgeprops={"edgecolor":"white"}
)

plt.title("Revenue Share by Product Category")
plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap:
import matplotlib.pyplot as plt
import seaborn as sns

numeric_columns = df[[
    "Age",
    "Quantity",
    "Price per Unit",
    "Total Amount"
]]

numeric_corr_matrix = numeric_columns.corr()

plt.figure(figsize=(7,5))

sns.heatmap(
    numeric_corr_matrix,
    annot = True,
    cmap = "coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of Sales Dataset")
plt.tight_layout()
plt.show()
```
